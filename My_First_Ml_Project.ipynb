```python
# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="darkgrid")
```

```python
# 2. CREATE DATASET

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120000, 135000, 142000, 160000, 155000, 175000],
    "Profit": [30000, 32000, 35000, 40000, 38000, 45000],
    "Orders": [1200, 1300, 1250, 1400, 1380, 1500]
}

df = pd.DataFrame(data)

print("📊 Dataset Preview")
print(df)
```

```python
# 3. BASIC INFO

print("\n Dataset Info")
print(df.info())

print("\n Missing Values")
print(df.isnull().sum())

print("\n Statistical Summary")
print(df.describe())
```

```python
# 4. LINE CHART - SALES TREND

plt.figure(figsize=(8,5))
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# 5. BAR CHART - PROFIT

plt.figure(figsize=(8,5))
plt.bar(df["Month"], df["Profit"])
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.show()
```

```python
# 6. BAR CHART - ORDERS

plt.figure(figsize=(8,5))
plt.bar(df["Month"], df["Orders"])
plt.title("Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Orders")
plt.show()
```

```python
# 7. CORRELATION HEATMAP (FIXED ERROR)

plt.figure(figsize=(5,4))

sns.heatmap(
    df.select_dtypes(include=["number"]).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# 8. PROFIT vs SALES (SCATTER PLOT)

plt.figure(figsize=(6,5))
plt.scatter(df["Sales"], df["Profit"])
plt.title("Sales vs Profit Relationship")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```
